# Optimización usando descenso de gradiente



Vamos a considerar el siguiente problema de minimización:
$$
\min_{\mathbf{x}\in \mathbb{R}^2} f(\mathbf{x}) \qquad (1)
$$
y el problema de maximización
$$
\max_{\mathbf{x}\in \mathbb{R}^2} f(\mathbf{x}) \qquad (2)
$$


Una solución aproximada $\widehat{\mathbf{x}}$ del problema de minimización puede obtenerse con el descenso de gradiente. Dicho algoritmo iterativo comienza con un valor inicial $\mathbf{x}^{(0)}=\mathbf{x}_0$ y para $k=1,2,3,...$  la iteración viene dada por:
$$
\mathbf{x}^{(k+1)}=\mathbf{x}^{(k)}-\alpha \nabla f(\mathbf{\mathbf{x}}^{(k)}),
$$

donde $\alpha$ es un parámetro positivo y los superíndices indican el número de iteración. Esta ecuación define una secuencia de vectores que pueden converger a la solución del problema de minimización, siempre que $\alpha$  sea adecuado. Para implementar este método, solo necesitamos el valor inicial $\mathbf{x}_0$ y el gradiente $\nabla f$.

El mismo algoritmo con una pequeña modificación permite resolver el problema de maximización.


1.1 (10 ptos) Considere la siguiente función:
$$
f(x,y)=\frac{1}{4}(x^4+y^4)-\frac{1}{3}(x^3+y^3)-(x^2+y^2)+4
$$

Halle el gradiente de $f$ y los candidatos a ser puntos extremos (son nueve). Luego encuentre la matriz Hessiana y úsela para clasificar los puntos como  mínimos, máximos o silla.

1.2 (12 ptos) Utilizando IA se obtuvo el siguiente código que busca resolver los problemas de optimización (1) y (2), usando la libreria $\texttt{scipy.minimize}$.

Usted debe intentar ajustar los valores iniciales para hallar **sólo** los valores extremos (máximos o mínimos) que encontró en el punto 1.1  y validar su clasificación numéricamente. Para cada punto debe existir evidencia de cual punto inicial le permitió hallar que punto extremo. Note que con el valor inicial dado en el código se encuentra un mínimo local pero no el máximo local. ¿Por qué cree usted que sucede esto?


**Nota:** Los puntos iniciales no pueden coincidir con los puntos candidatos a extremos.


```python
import numpy as np
from scipy.optimize import minimize

# Define the function
def f(vars):
    x, y = vars
    return (1/4)*(x**4 + y**4) - (1/3)*(x**3 + y**3) - (x**2 + y**2) + 4

# Initial guesses to explore multiple extrema
initial_guesses = [(-10, -10)]

results = []
for guess in initial_guesses:
    res = minimize(f, guess, method='BFGS')
    if res.success:
        results.append((res.x, res.fun))

# Print unique minima found
print("Local minima found:")
for point, value in results:
    print(f"At (x, y) = ({point[0]:.4f}, {point[1]:.4f}) -> f = {value:.4f}")

# To search for maxima, minimize -f
def neg_f(vars):
    return -f(vars)

results_max = []
for guess in initial_guesses:
    res = minimize(neg_f, guess, method='BFGS')
    if res.success:
        results_max.append((res.x, -res.fun))

print("\nLocal maxima found:")
for point, value in results_max:
    print(f"At (x, y) = ({point[0]:.4f}, {point[1]:.4f}) -> f = {value:.4f}")

```
