## Modelando la realidad: regresión

2. Vamos a describir una función afín muy comúnmente utilizada, especialmente cuando
el $n-$vector $x$ representa un vector de atributos. La función afín de $x$ dada por
$$\hat{y} = x^T \beta + v\qquad (1)$$
donde $\beta$   es un $n-$vector  y $v$ es un escalar, se denomina *modelo de regresión*. En este contexto,
las entradas de $x$ se denominan regresores, y $\hat{y}$ se denomina predicción, ya que el modelo de regresión es típicamente una aproximación o predicción de algún valor verdadero $y$, que se denomina variable dependiente, resultado o etiqueta.

El vector $\beta$  se denomina vector de pesos o vector de coeficientes, y el escalar $v$ se denomina compensación o intersección en el modelo de regresión. Juntos, $\beta$  y $v$ son llamados los parámetros en el modelo de regresión.

El símbolo $\hat{y}$ se usa en el modelo de regresión para enfatizar que es una estimación o
predicción de algún resultado $y$. Las entradas en el vector de peso tienen una interpretación simple:  $\beta_i$ es la cantidad por la cual $\hat{y}$ aumenta (si $\beta_i> 0$) cuando el atributo $i$ aumenta en uno (con todos los demás atributos iguales). Si  $\beta_i$ es pequeño, la predicción $\hat{y}$ no depende mucho del atributo $i$. La compensación $v$ es el valor de $\hat{y}$ cuando todas las características tienen el valor 0.
