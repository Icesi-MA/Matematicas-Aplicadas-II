## Método iterativo para un problema de **mínimos cuadrados**

Los métodos iterativos se utilizan a menudo para resolver problemas de mínimos cuadrados a gran escala. Vamos a considerar el siguiente problema de mínimos cuadrados:
$$
\min_{x \in \mathbb{R}^n} \|A\mathbf{x} - \mathbf{b}\|^2.
$$
Donde $A \in \mathbb{R}^{m \times n}$ y $\mathbf{b} \in \mathbb{R}^m$. Si $A$ tiene $n$ columnas linealmente independientes, sabemos que el problema tiene como única solución al vector, $\widehat{\mathbf{x}}=(A^\top A)^{-1}A^\top \mathbf{b}$, obtenido al resolver las ecuaciones normales.

Un solución aproximada a $\widehat{\mathbf{x}}$ puede obtenerse con un método iterativo. Dicho algoritmo puede comenzar con un valor inicial $\mathbf{x}^{(0)}=0$ y para $k=1,2,3,...$ se define una iteración
$$
\mathbf{x}^{(k+1)}=\mathbf{x}^{(k)}-\alpha A^\top\left(A \mathbf{x}^{(k)}-\mathbf{b}\right),\qquad (1)
$$

donde $\alpha$ es un parámetro positivo y los superíndices indican el número de iteración. Esta ecuación define una secuencia de vectores que convergen a la solución de mínimos cuadrados definida antes, es decir a $\widehat{\mathbf{x}}$, siempre que $\alpha$ no sea demasiado grande. Para implementar este método, solo necesitamos multiplicar los vectores dados por $A$ y por $A^\top$.

```python
import numpy as np

def iterative_least_squares(A, b, max_iter=500, tol=1e-6):
    """
    Solve the least squares problem ||Ax - b||_2 using an iterative method
    with step size alpha = 1 / ||A||_F^2.

    Parameters:
        A (np.ndarray): Matrix of shape (m, n).
        b (np.ndarray): Vector of shape (m,).
        max_iter (int): Maximum number of iterations.
        tol (float): Tolerance for stopping criterion.

    Returns:
        x (np.ndarray): Solution vector of shape (n,).
        residuals (list): List of residuals per iteration.
    """
    m, n = A.shape
    alpha = 1 / (np.linalg.norm(A, ord='fro') ** 2)    # Tamaño del paso
    x = np.zeros(n)  # Valor Inicial
    residuals = []

    for k in range(max_iter):
        residual = A @ x - b
        gradient = A.T @ residual  # Gradiente de ||Ax - b||^2 es 2A^T(Ax - b)
        x_new = x - alpha * gradient

        # Calcule el error residual
        residual_norm = np.linalg.norm(A @ x_new - b)
        residuals.append(residual_norm)

        # Convergencia para mínimos cuadrados: ||A^T(Ax-b)|| debe ser pequeña.
        gradient_new = A.T @ (A @ x_new - b)
        x = x_new
        if np.linalg.norm(gradient_new) < tol:
            break

    return x, residuals

# Example usage:
if __name__ == "__main__":
    # Bloque a modificar para ser ajustado acorde al enunciado 1.2
    m, n = 200, 100
    A = np.random.randn(m, n)
    b = np.random.randn(m)

    # Calcule una solución de referencia estable del problema de mínimos cuadrados.
    x_true = np.linalg.lstsq(A, b, rcond=None)[0]

    # Resuelva usando el método iterativo
    x_est, residuals = iterative_least_squares(A, b, max_iter=500)

    print("Estimated x:", x_est)
    print("True x:", x_true)
    print("Final residual:", residuals[-1])
```
