# Modelando con EDOs: Método de Runge-Kutta y comparación con Euler

Consideremos la EDO general

$$
\frac{dx}{dt}=f(t,x),\qquad x(0)=x_0.
$$

El **método de Runge-Kutta** (RK) se define con la siguiente iteración

$$
x_{n+1} = x_n + \tfrac{h}{2}(k_1 + k_2).
$$


donde la primera pendiente (Euler) es:

$$
k_1 = f(t_n, x_n).
$$

y luego estimamos un valor intermedio usando $k_1$:

$$
k_2 = f\left(t_n + h, \; x_n + h k_1 \right).
$$

Este método es **explícito** y de **orden 2**, más preciso que Euler.
