## Modelando funciones con sumas infinitas: series de Fourier

1. La integración explícita de funciones suele ser imposible o muy compleja, por lo que se deben utilizar métodos numéricos en su lugar. Vimos por ejemplo que la regla del trapecio y la regla de Simpson son métodos comunes para aproximar integrales. Cada método tiene un orden de precisión que depende de la aproximación del área debajo de la función. Por otro lado, algunas funciones pueden expresarse como una suma infinita de polinomios (es decir, la serie de Taylor). Otras funciones, en particular las periódicas, pueden escribirse como una suma infinita de ondas seno y coseno. Para estas funciones,

$$f(x)= \frac{A_0}{2}+\sum_{n=1}^{\infty}A_n\cos{(nx)} + B_n\sin{(nx)}\quad (F)$$,

Se puede demostrar que los valores de $A_n$ y $B_n$ son calculados utilizando las siguientes fórmulas:

  $$A_n= \frac{1}{\pi} \int_{-\pi}^{\pi} f(x) \cos{(nx)}\, dx\quad (1)$$,
  $$B_n= \frac{1}{\pi} \int_{-\pi}^{\pi} f(x) \sin{(nx)}\, dx\quad (2)$$

Al igual que con las series de Taylor, las funciones se pueden aproximar truncando la serie de Fourier en algún punto $n = N$. Las series de Fourier se pueden utilizar para aproximar algunas funciones particularmente complicadas, como la función escalonada, y forman la base de muchas aplicaciones de ingeniería, como el procesamiento de señales.

```python
def plot_results(f, N):
    """
    Dibuja la serie de Fourier de una funcion f usando la regla  trapezoidal y de  Simpson.

    Parameters:
        f (funcion)
        N (int): Numero de terminos de la serie de  Fourier .
    """
    x = np.linspace(-np.pi, np.pi, 10000)  #

    # Inicializar la serie de  Fourier
    y_trap = np.zeros(len(x))
    y_sim = np.zeros(len(x))

    # Calcule A0 and B0
    A0_trap, B0_trap = fourier_trap(f, 0)
    A0_sim, B0_sim = fourier_sim(f, 0)

    # Añada  A0  (B0 es zero para n=0)
    y_trap += A0_trap / 2
    y_sim += A0_sim / 2

    # Calcule los terminos de orden superior
    for n in range(1, N):
        An_trap, Bn_trap = fourier_trap(f, n)
        An_sim, Bn_sim = fourier_sim(f, n)

        y_trap += An_trap * np.cos(n * x) + Bn_trap * np.sin(n * x)
        y_sim += An_sim * np.cos(n * x) + Bn_sim * np.sin(n * x)

    # Gráfica
    plt.figure(figsize=(12, 6))

    # Regla rapezoidal
    plt.subplot(1, 2, 1)
    plt.plot(x, f(x), label='Analitica')
    plt.plot(x, y_trap, label='Trapezoidal')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.grid()
    plt.legend()
    plt.title(f'{N}th orden aproximación de  Fourier (Trapezoidal)')

    # Regla de  Simpson
    plt.subplot(1, 2, 2)
    plt.plot(x, f(x), label='Analitica')
    plt.plot(x, y_sim, label="Simpson ")
    plt.xlabel('x')
    plt.ylabel('y')
    plt.grid()
    plt.legend()
    plt.title(f'{N}th orden aproximación de  Fourier (Simpson)')

    plt.tight_layout()
    plt.show()

# Define la funcion de cada caso
f = lambda x: np.sin(np.exp(x)) # f2 = np.mod(x, np.pi/2) #f3 (x > -np.pi/2) & (x < np.pi/2)

# Numero de terminos de Fourier
N = 2  #  5, 20

# Graficos
plot_results(f, N)
```
