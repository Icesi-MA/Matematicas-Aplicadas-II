Parece  poco realista, pero eso es exactamente lo que hace el  método a estudiar ahora.

  **Método de Euler (implicito)**
  Queremos resolver la ecuación diferencial
  $$\frac{dx}{dt}=f(x,t),\quad x(0)=x_0$$Para ello aproximamos la derivada mirando hacia adelante en el tiempo, esto es, $$\frac{x(t+h)-x(t)}{h}\approx f(x(t+h),t+h),\quad x(0)=x_0$$ lo cual se reescribe como $$x(t+h)\approx x(t) + h f(x(t+h),t+h)$$ o por simplicidad $$x_{n+1}=x_n+hf(x_{n+1},t_{n+1}).$$
  Siempre sabremos el valor de $t+h$  y  el valor de $t$, pero no sabemos el valor de $x(t+h)$. De hecho, eso es exactamente lo que queremos. El mayor problema es que $x(t+h)$  aparece en ambos lados de la ecuación.
  
  Este método se llama el método de Euler hacia atrás y se conoce como un método implícito ya que no calcula $x(t+h)$ explícitamente, sino que hace algún cálculo intermedio para hallar $x(t+h)$. La ventaja (habitual) de un método implícito como el método de Euler hacia atrás es que puede dar muchos menos pasos con una pérdida de precisión razonablemente pequeña. La idea es estudiar precisamente esta característica.
