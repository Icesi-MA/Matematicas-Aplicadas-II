 # Matriz de covarianza

2. (10 ptos)  Considere los vectores del punto anterior y defina la matriz $A=[a_1\;a_2\;a_3]$.

  a. Calcule el $3$-vector $\mu$ de las medias de las columnas, es decir,  $\mu_i = avg(a_i), i = 1,2,3$. Luego calcule los vectores $\tilde{a}_i=a_i-\mu_ie_i$ y defina $\tilde{A}$ como la matriz $\tilde{A}=[\tilde{a}_1\;\tilde{a}_2\;\tilde{a}_3]$

  c. Calcule la matriz de *covarianza*  $\Sigma = (\frac{1}{3})\tilde{A}^T\tilde{A}$, de los vectores  $a_1,a_2,a_3$ y verifique que

  $$\Sigma_{ij}=\begin{cases}
\mathbf{std}(a_i)^2, & \text{ si } i=j\\
\mathbf{std}(a_i)\,\mathbf{std}(a_j)\,\rho_{ij}, & \text{ si } i\not=j\end{cases}$$

donde $\rho_{ij}$ es el coeficiente de correlación de $a_i$ y $a_j$.
