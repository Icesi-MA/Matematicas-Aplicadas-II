## Diferenciación de funciones  en varias variables

> Presentamos ahora la noción de derivada de funciones de varias variables. Empezamos con funciones de dos variables y luego extendemos a varias variables usando notación vectorial y matricial encontramos que muchos de los conceptos
y los resultados parecen familiares a los de una variable.

Recuerde que para una función $f$ de una sola variable $x$, decimos que $f$ es diferenciable en $a$ si f es localmente lineal en $a$. Es decir, hay una constante $m$ tal que para todo $h$ suficientemente cercano a cero el cambio en $f$ en $a$,
$$f (a+h)− f (a),$$
está bien aproximado por $mh$. Por "bien aproximado" queremos decir que la diferencia
entre $f(a+h)− f (a)$ y $mh$ es pequeña en comparación con $h$ cuando $h$ es pequeña. Es decir,
$\frac{f (a+h)− f (a) −mh}{h}$

tiende a cero cuando $h$ tiende a cero. En este caso decimos que $f$ es diferenciable en $a$ y
escribimos
$$\lim_{h→0} \frac{f (a+h)− f (a)}{h}=m$$

El número $m$ se llama la derivada de $f$ en $a$, denotada $f'(a)$, y la aproximación lineal
de $f$ en $a$ es
$$f(a)+ f'(a)(x−a).$$
Ahora, extendamos la noción de "linealidad local" a las funciones de $\mathbb{R}^{2}→\mathbb{R}$. Primero
recordamos  que una función lineal de $\mathbb{R}^{2}→\mathbb{R}$ es una función de la forma $\ell(h, k) = ph+qk$,
donde $p$ y $q$ son algunos números. Recordamos la definición de norma de un vector
$$||(h, k)|| = \sqrt{h^2 +k^2}.$$

**Definición:** Una función $f$ definida en un dominio de $\mathbb{R}^{2}$ es diferenciable en $(a,b)$ si

$$\lim_{||(h, k)|| →0} \frac{f (a+h,b+k)− f (a,b)-\ell(h,k)}{||(h, k)||}=0$$










Veamos esto en mayour detalle. Supongamos que $f$ es derivable en $(a,b)$ y $(h, k) = ph + qk$. Hagamos $k = 0$ en $\ell(h, k) = ph+qk$ entonces usted puede decir que el límite
$$\lim_{h →0} \frac{f (a+h,b)− f (a,b)-ph}{h}=???,$$
luego $f (x,b)$ es una función diferenciable de una sola variable $x$ en $a$, y
$$\lim_{h →0} \frac{f (a+h,b)− f (a,b)-ph}{h} = p$$
El número $p$ se llama la derivada parcial de $f$ con respecto a $x$ en $(a,b)$ y se
denota
$$\frac{∂ f}{∂x}(a,b)$$ o $f_x(a,b)$.
Por lo tanto $$\frac{∂ f}{∂x}(a,b)$$ se encuentra manteniendo y igual a la constante $b$ y derivando $f (x,b)$
con respecto a $x$ en $a$. De manera similar, si hacemos $h = 0$ en (1), vemos que si $f$ es
diferenciable en $(a,b)$ entonces $f (a, y)$ es una función diferenciable de $y$ en $b$ y
$$\lim_{k →0} \frac{f (a,b+k)− f (a,b)-qk}{k},$$

así que
$$ \lim_{k →0} \frac{f (a,b+k)− f (a,b)}{k}= q$$
El número $q$ se llama derivada parcial de $f$ con respecto a $y$ en $(a,b)$ y es
denotado
$$\frac{∂ f}{∂y}(a,b)$$ o $f_y(a,b)$.
Esto muestra que si $f$ es derivable en $(a,b)$ entonces tiene derivadas parciales $f_x(a,b)$
y $f_y(a,b)$ allí, y la aproximación lineal de $f$ en $(a,b)$ es
$$L(x, y) = f (a,b)+ f_x(a,b)(x−a)+ f_y(a,b)(y−b)=f(a,b)+\ell(x-a,y-b).$$

**Teorema** Si $f:\mathbb{R}^{2}→\mathbb{R}$ es diferenciable en $(a,b)$ entonces es continua en  $(a,b)$.

Las reglas de derivación parcial son equivalentes a las reglas de derivación de derivación ordinaria



1.   $(f+g)_x=f_x+g_x$, $(f+g)_y=f_y+g_y$
2.   $(fg)_x = f_xg+fg_x$, $(fg)_y = f_yg+fg_y$

**Teorema.** Si las derivadas parciales de $f$ son continuas en un dominio que contiene $(a,b)$, entonces $f$ es diferenciable en $(a,b)$.

## Plano tangente y derivadas parciales

La interpretación geométrica de la derivada $f'(a)$ de una función $f$ de una sola variable es la pendiente de la recta
$y = f(a)+ f'(a)(x−a)$
tangente a la gráfica de la función. Hay una interpretación geométrica similar de la
derivadas parciales de una función de dos variables. Supongamos que $f$ es derivable en $(a,b)$,
con derivadas parciales $f_x(a,b)$ y $f_y(a,b)$ en $(a,b)$. La interpretación geométrica de
aproximación lineal es que la gráfica de
$$z = L(x, y) = f(a,b)+ f_x(a,b)(x−a)+ f_y(a,b)(y−b)$$
es el plano tangente a la gráfica de $f$ en el punto  $(a,b, f (a,b)$ . Reescribiendo la ecuación
$$f_x(a,b)(x−a)+ f_y(a,b)(y−b)+(−1)(z − f (a,b))  = 0$$
vemos que una normal al plano tangente es el vector

$$N = ( f_x(a,b), f_y(a,b),−1).$$
Si $f$ es diferenciable en $(a,b)$ decimos que $N$ es normal a la gráfica de $f$ en $(a,b, f (a,b)) $.
