#Extremos de funciones de varias variables

> Tal como en cálculo de una variable, las funciones de varias variables pueden alcanzar valores extremos en ciertos puntos y nos interesa tanto encontrarlos como clasificarlos.

**Definición. Puntos extremos.** Sea $f :D\subset \mathbb{R}^n→\mathbb{R}$. Decimos que $f$ tiene un mínimo local $f (A)$ en un punto $A$ en $D$ si hay una vecindad abierta en $\mathbb{R}^n$ con centro en $A$ de modo que para cada
punto $X$ en $D$ que está en la vecindad, $$f (X) ≥ f (A)$$ $f$ tiene un máximo local $f (A)$ en un punto $A$ en $D$ si hay una vecindad abierta en $\mathbb{R}^n$
con centro en $A$ de modo que para cada punto $X$ en $D$ que está en la bola,
$$f (X) ≤ f (A)$$
Un punto que es un *máximo local o un mínimo local* se llama punto local
extremo de $f$ en $D$. Decimos que un máximo local o un mínimo local es estricto si la desigualdad estricta correspondiente $f (X) > f (A)$ o $f (X) < f (A)$ se cumple excepto para $X = A$.



**Teorema (criterio de la primera derivada)** Sea $f$ una función diferenciable  $f :D\subset \mathbb{R}^n→\mathbb{R}$. Supongamos
$f$ tiene un extremo local $f (A)$ en el punto interior $A$. Entonces las primeras  derivadas parciales  de $f$ son cero en $A$, es decir,
$$∇f(A) = 0.$$

Recordemos que el criterio de la primera derivada para funciones diferenciables proviene del hecho geométrico derivado del comportamiento de los planos tangentes a la gráfica de una función donde hay máximos o mínimos. Los planos tangentes en dichos puntos son horizontales.

**Ejemplo.** Considere las funciones $f(x,y)=x^2+y^2$, $g(x,y)=-x^2-y^2$ y $h(x,y)=x^2-y^2$. Encuentre los candidatos a  puntos extremos e intente clasificarlos.

**Definición (Matriz Hessiana)** Dada una función $C^2$  $f :\mathbb{R}^n→\mathbb{R}$ la matriz $n \times n$ de segundas derivadas parciales en $(x_1, . . . , x_n)$

$$H(x)=\begin{bmatrix}
f_{x_1x_1} & f_{x_1x_2} & \dots & f_{x_1x_n}\\
f_{x_2x_1} & f_{x_2x_2} & \dots & f_{x_2x_n}\\
\vdots & \vdots & \ddots & \vdots\\
f_{x_nx_1} & f_{x_nx_2} & \dots & f_{x_nx_n}
\end{bmatrix}
$$
es llamada la matriz *Hessiana* de $f$ en $x$.



**Teorema (Criterio de la segunda derivada en $\mathbb{R}^{2}$)** Supongamos que $f$ es una función $C^2$ en el interior de $D ⊂ \mathbb{R}^{2}$, y que en algún punto interior $(c,d)$ de $D$,
$$∇f (c, d) = 0.$$

Sea $H(c,d)$ la matriz hessiana de segundas derivadas parciales de $f$ en $(c,d)$,
$$H(c,d)=\begin{bmatrix}
f_{x_1x_1}(c,d) & f_{x_1x_2}(c,d)\\
f_{x_2x_1}(c,d) & f_{x_2x_2}(c,d)
\end{bmatrix}
$$

(a) Si $H(c,d)$ es definida positiva, entonces $f$ tiene un mínimo local estricto en $(c,d)$. Esto es, si $f_{xx}>0$ y $\det (H)>0$.

(b) Si $H (c,d)$ es definida negativa, entonces $f$ tiene un máximo local estricto en $(c,d)$. Esto es, si $f_{xx}<0$ y $\det (H)>0$.

(c) Si $H (c,d)$ es indefinida, entonces $f$ no tiene un máximo local ni un mínimo local en ($c, d)$. Esto es, si $\det (H)<0$. Estos puntos se denominan *puntos silla*.

**Ejemplo** Encuentre y clasifique los puntos extremos locales de $f(x,y)=2x^2-xy+y^4$.

Es de anotar, que para funciones de $n$ variables se pueden extender estos resultados pero utilizan conceptos de algebra lineal, *valores y vectores propios*, que no han sido abordados en este curso por lo cual se estudiarán en el futuro.

#Extremos de funciones de varias variables sobre conjuntos de nivel

El teorema del valor extremo garantiza que una función continua $f$ de un
conjunto cerrado y acotado $D ⊂ \mathbb{R}^{n}$ a $ \mathbb{R}$ tiene un valor máximo absoluto y un mínimo absoluto. Hemos visto que  si una función $C^1$ $f:D ⊂ \mathbb{R}^{n}→\mathbb{R}$ tiene un
máximo o mínimo en un punto interior $A$ entonces $∇f (A) = 0$. Este “ criterio de la primera derivada” nos da una manera de encontrar candidatos para los extremos que ocurren en el interior de $D$.

Ahora vamos a dar un resultado que permite identificar candidatos a extremos de una función $f$ en un conjunto de nivel. Enunciamos el teorema para funciones de $\mathbb{R}^2$ a $\mathbb{R}$ pero un resultado análogo es cierto en cualquier otra dimensión.


**Teorema (Multiplicadores de Lagrange)**. Sean $f$ y $g$ funciones $C^1$ de un conjunto abierto en $\mathbb{R}^2$ a $\mathbb{R}$, y denotamos por $S$ un conjunto de nivel (restricción) $g(x, y) = c$. Sea $P$ un punto de $S$ tal que

 (a) $∇g(P) \not=  0$, y

(b) $f$ tiene un extremo local en $P$. Entonces existe un número $λ$ tal que

$$∇f (P) = λ∇g(P). \qquad (1)$$

El número $\lambda$ se conoce como multiplicador de Lagrange.

```python
import numpy as np
import matplotlib.pyplot as plt
xdomain = np.linspace(-4,4,100)
ydomain = np.linspace(-4,4,100)
X, Y = np.meshgrid(xdomain,ydomain)
f = lambda x, y: -(x+y)
#f = lambda x, y: np.sin(x)*np.exp(-np.sqrt(x**2+y**2))
plt.contour(X,Y,f(X,Y))
plt.grid()
plt.show()
```

Consideremos el siguiente problema:

```python
def f(x):
    return -(x[0] + x[1])  # maximize (x0 + x1)

cons = ({
    'type': 'eq',
    'fun': lambda x: x[0]**2 + x[1]**2 - 2   # scalar constraint: x0^2 + x1^2 = 2
})

# Initial guess
x0 = np.array([0, 1])
```

```python
import numpy as np
from scipy.optimize import minimize

def callback(xk):
    print(f"Current iterate: {xk}, Objective value: {f(xk):.6f}")

res = minimize(f, x0, method='SLSQP', constraints=cons, callback=callback, options={'disp': True})

print("\nOptimization complete.")
print(f"Optimal x: {res.x}")
print(f"Final objective value: {res.fun}")
```

# Teorema de Taylor en varias variables

### 🧮 **Fórmula de Taylor de Primer Orden**

Sea $f : \mathbb{R}^n \to \mathbb{R}$ una función diferenciable.
Entonces, alrededor de un punto $\mathbf{a} \in \mathbb{R}^n$:

$$
f(\mathbf{x}) \approx f(\mathbf{a}) + \nabla f(\mathbf{a}) \cdot (\mathbf{x} - \mathbf{a})
$$

Es decir:

$$
f(\mathbf{x}) = f(\mathbf{a}) + \sum_{i=1}^n
\frac{\partial f}{\partial x_i}(\mathbf{a}) (x_i - a_i) + R_1(\mathbf{x})
$$

donde $R_1(\mathbf{x})$ es el **término de resto de primer orden**, que tiende a cero más rápidamente que $|\mathbf{x} - \mathbf{a}|$ cuando $\mathbf{x} \to \mathbf{a}$.

---

### 🧮 **Fórmula de Taylor de Segundo Orden**

Si $f$ tiene derivadas parciales segundas continuas cerca de $\mathbf{a}$, entonces:

$$f(\mathbf{x}) \approx f(\mathbf{a}) + \nabla f(\mathbf{a}) \cdot (\mathbf{x}- \mathbf{a}) \frac{1}{2} (\mathbf{x} - \mathbf{a})^T H_f(\mathbf{a}) (\mathbf{x} - \mathbf{a})$$

o, en forma desarrollada por componentes:

$$
f(\mathbf{x}) = f(\mathbf{a})+\sum_{i=1}^n \frac{\partial f}{\partial x_i}(\mathbf{a})(x_i - a_i)
+ \frac{1}{2} \sum_{i=1}^n \sum_{j=1}^n
  \frac{\partial^2 f}{\partial x_i \partial x_j}(\mathbf{a})(x_i - a_i)(x_j - a_j)
+ R_2(\mathbf{x})
$$

donde $H_f(\mathbf{a})$ es la **matriz Hessiana** de $f$ en $\mathbf{a}$, y $R_2(\mathbf{x})$ es el **término de resto de segundo orden**.

## 📘 Descenso del Gradiente

El **descenso del gradiente** (*Gradient Descent*) es uno de los algoritmos más fundamentales en **optimización numérica** y **aprendizaje automático**.
Sirve para encontrar el mínimo (local o global) de una función. Supongamos que tenemos una función ($ f(\mathbf{x}) $) que queremos **minimizar**.
Partimos de un valor inicial ($ \mathbf{x}_0 $) y avanzamos iterativamente en la dirección **opuesta al gradiente**, es decir, “cuesta abajo”:

$$
\mathbf{x}_{k+1} = \mathbf{x}_k - \alpha \nabla f(\mathbf{x}_k)
$$

donde:

* $ \mathbf{x}_k $: punto actual (iteración $k$)
* $ \alpha $: tasa de aprendizaje (*learning rate* o tamaño de paso)
* $ \nabla f(\mathbf{x}_k) $: gradiente (vector de derivadas parciales)


### ⚙️ **Ejemplo en 1D**

Queremos minimizar:

$$
f(x) = x^2 + 3x + 2
$$

Su derivada (el gradiente) es:

$$
f'(x) = 2x + 3
$$

---


### 📊 **Observaciones**

* En cada paso, el algoritmo actualiza $ x $ restando una fracción del gradiente $ f'(x) $.
* Si la tasa de aprendizaje $ \alpha $ es **demasiado grande**, puede **sobrepasar** el mínimo.
* Si $ \alpha $ es **demasiado pequeña**, el algoritmo será **lento** en converger.

---

### 🔍 **Mínimo Analítico**

Podemos hallar el mínimo resolviendo $ f'(x) = 0 $:

$$
2x + 3 = 0 \Rightarrow x = -1.5
$$

El algoritmo debería converger hacia $ x = -1.5 $, el punto donde la función alcanza su valor mínimo.


```python
### 🧮 **Implementación en Python**

import numpy as np
import matplotlib.pyplot as plt

# Definimos la función y su gradiente
def f(x):
    return x**2 + 3*x + 2

def grad_f(x):
    return 2*x + 3

# Parámetros del descenso del gradiente
alpha = 0.1   # tasa de aprendizaje
x = 5.0       # valor inicial
iterations = 20

# Para almacenar los valores
x_values = [x]
f_values = [f(x)]

# Bucle principal
for i in range(iterations):
    x = x - alpha * grad_f(x)
    x_values.append(x)
    f_values.append(f(x))
    print(f"Iter {i+1}: x = {x:.4f}, f(x) = {f(x):.4f}")

# Visualización
x_plot = np.linspace(-5, 5, 100)
plt.plot(x_plot, f(x_plot), label='f(x)')
plt.scatter(x_values, f_values, color='red', label='Pasos del GD')
plt.title("Descenso del Gradiente en $f(x) = x^2 + 3x + 2$")
plt.xlabel("$x$")
plt.ylabel("$f(x)$")
plt.legend()
plt.grid(True)
plt.show()

```
