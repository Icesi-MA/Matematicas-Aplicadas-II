#Aproximación de integrales

**Reglas del punto medio, trapecio y Simpson**

En general, las fórmulas de aproximación para integrales vienen determinadas por la escogencia del punto de prueba $x^*$. Son de la forma $$ A_{approx}=h\sum_{i=0}^n f(x^*)$$
Por ejemplo, la regla de la mano derecha $$A_R=h\sum_{i=0}^n f(a+ih)$$ y la regla de la mano izquierda $$A_L=h\sum_{i=0}^{n-1} f(a+ih)$$ nos permiten aproximar el valor de una integral $$\int_a^b f(x)\;dx$$
donde $h=\frac{b-a}{n}$.

Sin embargo, hay aproximaciones mas precisas como la del punto medio,
$$A_M=h\sum_{i=0}^n f(a+(i-\frac{1}{2}h),$$
la del trapezoide, que se puede calcular como $$A_T=\frac{A_L+A_R}{2},$$ y la de Simpson $$A_S=\frac{1}{3}A_T+ \frac{2}{3}A_M.$$ A continuación vemos implementaciones numéricas de ellas.

```python
import numpy as np

def midpoint(f, a, b, n):
    h = float(b-a)/n
    result = 0
    for i in range(n):
        result += f(a + (i - 0.5)*h)
    result *= h
    return result

def trapezoidal(f, a, b, n):
    h = float(b-a)/n
    result = 0.0
    for i in range(1, n):
        result += f(a + i*h)
    result *= h
    return result

def simpson(f, a, b, n):
    # calculating step size
    h = float(b - a) / n
    x = np.linspace(a, b, n)
    # Finding sum
    result = 0

    for i in range(n-1):
        #a0 = a + (i-1)*h
        #a1 = a + (i-0.5)*h
        #a2 = a + i*h
        result += f(x[i])+4*f((x[i]+x[i+1])/2)+f(x[i+1])

    result *= h/6

    return result
```

```python
def g(x):
  return np.log(x)

a = 1
b = 2
n = 100 # Number of subdivisions

simpson_result = simpson(g, a, b, n)

print(f"La integral de ln(x) de {a} a {b} usando la regla de Simpson con {n} subdivisiones es: {simpson_result}")
```

```python
def f(x):
  if abs(x)>0:
    fx = np.log(x)/x
  else:
    fx=0

  return fx

def intf(x):
  value = 0.5*np.log(x)**2

  return value
n = 100
md = midpoint(f, 1, 2, n)

tr = trapezoidal(f, 1, 2, n)

si = (2/3)*md + (1/3)*tr

si3 = simpson(f,1, 2, n)

exact = intf(2) - intf(1)

print("Integrales aproximadas por el método de punto medio: %e " % (md) + "trapecio: %e " % (tr)+ " simpson: %e " % (si) + " simpson 1/3: %e " % (si3)  )
print("Valor exacto: %e " % (exact))
print("Errores relativos punto medio: %e" % (np.abs((exact-md)/exact)) + " trapecio: %e " % (np.abs((exact-tr)/exact)) + " simpson: %e " % (np.abs((exact-si)/exact)) + " simpson 1/3: %e " % (np.abs((exact-si3)/exact)))
```

# Métodos de Integración

**Integración por sustitución**

A partir de la regla de la cadena para una función $F$ (que puede ser la antiderivada de una función $f$) tenemos

$$\left(F(g(x))\right)' =  F'(g(x))g'(x),$$

luego podemos deducir la siguiente fórmula para integrar.

**Teorema (Cambio de variable)** Sea $g$ una función con derivada continua en un intervalo $[a,b]$, y sea $f$ continua en el intervalo que contiene al rango de $g$. Entonces

$$\int f(g(x))g'(x)\, dx=\int f(u) \frac{du}{dx} dx=\int f(u)\,du.$$

o equivalentemente en el caso de una integral definida

$$ \int_{g(a)}^{g(b)} f(u)\,du=\int_a^b f(g(x))g'(x)\, dx.$$

**Integración por partes**

Ahora recordemos la regla del producto para derivadas

$$(f g ) ' =f'g+fg',$$

luego si integramos sobre un intervalo $[a,b]$ y suponiendo como antes que $f'$ y $g'$ son continuas, obtenemos  

**Teorema (Integral por partes)** Sean $f'$ y $g'$  continuas en un intervalo $[a,b]$. Entonces

$$\int f'(x)g(x)\, dx=f(x) g(x)-\int f(x)g'(x)\,dx.$$
