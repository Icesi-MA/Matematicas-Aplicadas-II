#Regla de la cadena

> Hay muchos tipos de funciones de varias variables veremos la formulación de la regla de la cadena en cada una de sus formas.

**Teorema. Regla de la cadena 1 (para curvas).** Sea $f :\mathbb{R}^2→\mathbb{R}$ continuamente
diferenciable en un conjunto abierto $U$, y sea $X(t) = (x(t),y(t))$  una función diferenciable
 de un intervalo abierto $I$ a $U$. Entonces la composición $f(x(t), y(t))$ es una función diferenciable de $I$ a $\mathbb{R}$ y

 $$\frac{d}{dt}f(x(t), y(t))=\frac{\partial f}{\partial x}\frac{dx}{dt}+\frac{\partial f}{\partial y}\frac{dy}{dt},$$
 o equivalentemente

 $$\frac{d}{dt}f(x(t), y(t))=\nabla f(x(t),y(t))\cdot X'(t).$$

Para el caso $n-$dimensional tendríamos $f :\mathbb{R}^n→\mathbb{R}$  y sea $X(t) = (x_1(t),...,x_n(t))$, de donde

$$\frac{d}{dt}f(X(t))=\frac{\partial f}{\partial x_1}\frac{dx_1}{dt}+...+\frac{\partial f}{\partial x_n}\frac{dx_n}{dt},$$
 o equivalentemente

 $$\frac{d}{dt}f(X(t))=\nabla f(X(t))\cdot X'(t).$$

**Ejemplo** Sea $X(t)=(\cos t,\sin t,t)$ la posición de una partícula en $\mathbb{R}^3$ en el tiempo $t$. Sea $f(x,y,z)=e^{-x^2-y^2-z^2}$ representa la temperatura de la partícula en el punto $(x,y,z)$. Entonces la función compuesta

$$f(X(t))=e^{-(1+t^2)}.$$

es la temperatura de la partícula en el tiempo $t$. Halle la tasa a la cual la temperatura de la partícula cambia.

**Derivada direccional** La tasa a la cual $f :\mathbb{R}^n→\mathbb{R}$ cambia cuando nos movemos de un punto $P$ a lo largo de la linea recta $X(t)=P+tV,\;-h\leq t \leq h $ es

$$ \lim_{t \rightarrow 0} \frac{f(X(t))-f(X(0))}{t-0}=\frac{d}{dt} f(X(t))|_{t=0}.$$

Por la regla de la cadena,

$$\frac{d}{dt} f(X(t))=\nabla f(X(t))\cdot X'(t)$$



**Definición.** Sea la función $f:\mathbb{R}^{n}→\mathbb{R}$ y $V$ un vector unitario. Denominamos
$$D_Vf(P)=∇f(P) \cdot V$$
a la derivada diraccional de $f$ en el punto $P$ en la dirección de $V$.

**Ejemplo** Sea $f(x,y,z)=x+y^2+z^4$. Encuentre la dirección y la magnitud de la derivada direccional mas grande en el punto $P=(3,2,1)$.

#Derivadas parciales de orden superior

**Definición**. Una función f definida en un disco abierto en el plano $x, y$ es dos veces
continuamente diferenciable en el disco si existen sus derivadas parciales $f_x$ y $f_y$
y tienen derivadas parciales continuas. Las derivadas parciales de $f_x$ y $f_y$ se denotan
 como $f_{xx}, f_{xy}$ y $f_{yx}, f_{yy}$ y se llaman segundas derivadas parciales.

Una función diferenciable dos veces continuamente se llama función $C^2$.

**Teorema.** Para una función $f$ continuamente diferenciable dos veces  en un disco abierto
 en el plano $x, y$, las segundas derivadas parciales mixtas son iguales:
$$f_{xy} = f_{yx}.$$

**Ejercicio.** Encuentre todas las segundas derivadas parciales de la función $f(x,y,z)=e^{xy}+z\cos x$.
