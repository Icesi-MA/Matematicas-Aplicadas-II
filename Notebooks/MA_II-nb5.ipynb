## Funciones  en varias variables

> Antes mencionamos que la descripción matemática de ciertos fenómenos del mundo real requiere de una colección de números, y que tales situaciones se abstraen en el concepto de  vector o de matriz. Ahora, vamos a construir otro objeto matemático cuyo dominio involucra tanto vectores como matrices para describir  situaciones, fenómenos o procesos complejos que son de interés para la ciencia y la ingeniería.

Una función  que mapea $n-$vectores en los reales, en otras palabras una función de valor real de $n-$vectores, es un ejemplo de un función en varias variables. Si $x$ es un $n-$vector, entonces $f(x)$, que es un escalar, denota el valor de la función $f$ en $x$. (En la notación $f(x)$, se hace referencia a $x$ como el argumento de la función).
También se puede interpretar $f$ como una función de $n$ argumentos escalares, en cuyo caso escribimos $f(x)$ como

$$f(x)=f(x_1,x_2,...,x_n).$$


Para describir una función $f :\mathbb{R}^n→\mathbb{R}$ , tenemos que especificar cuál es su valor para cualquier
posible argumento $x\in\mathbb{R}^n$. Por ejemplo, podemos definir una función $\mathbb{R}^{4}→\mathbb{R}$ por

$$f(x)=x_1+x_2-x_4^2.$$

Por otro lado, muchas funciones no están dadas por fórmulas o ecuaciones. Como ejemplo, supongamos $f:\mathbb{R}^{3}→\mathbb{R}$ es la función que da la elevación (fuerza vertical hacia arriba) en un determinado avión, en función del $3-$vector $x$, donde $x_1$ es el ángulo de ataque del avión (es decir, el ángulo entre el cuerpo del avión y su dirección de movimiento), $x_2$ es la velocidad del aire, y $x_3$ es la densidad del aire.


## Otros ejemplos



1.   *Función producto escalar (interno)*. Suponga que $a$ es un $n-$vector. Podemos definir la fución de valor real
$$f(x)=a^Tx=a_1x_1+a_2x_2+...+a_nx_n\qquad (1)$$ parar cualquier vector $x$. Esta función da el producto interno de su argumento el $n-$vector $x$ con algún $n-$vector fijo $a$. También podemos pensar en $f$ como formando una suma ponderada de
los elementos de $x$; los elementos de $a$ dan los pesos utilizados en la formación de la suma ponderada.

2. *Superposición y linealidad*. La función de producto interior $f$ definida en (1) satisface
la propiedad, usted debe demostrarlo,

$$ f(\alpha x+\beta y)=\alpha f(x)+\beta f(y)\qquad (2)$$

para cualquier par de vectores $x$ y $y$, y todos los escalares $\alpha$ y $\beta$. Esta propiedad se denomina *principio de superposición*. Toda función que satisface la propiedad de superposición se denomina *lineal*. Si una función $f$ es lineal, la superposición se extiende a combinaciones lineales de cualquier
número de vectores:

$$f(\alpha_1x_1+...+\alpha_k x_k)=\alpha_1f(x_1)+...+\alpha_kf(x_k)$$

3. La igualdad de superposición (s) a veces se descompone en dos propiedades, una que involucra el producto escalar-vector y otra que involucra la suma de vectores. Una función $f : \mathbb{R}^n→\mathbb{R}$ es lineal si satisface las siguientes dos propiedades
  - Homogeneidad. Para cualquier $n-$vector $x$ y cualquier escalar $\alpha$, $f(\alpha x) = \alpha f(x)$.
  - Aditividad. Para cualquier $n-$vector $x$ y $y$, $f(x + y) = f(x) + f(y)$.
4. *Funciones afines*. Una función lineal más una constante se llama función afín. Una función $f : \mathbb{R}^n→\mathbb{R}$ es afín si y solo si puede expresar como $f(x) = a^T x + b$ para algún $n-$vector $a$ y un escalar $b$. Por ejemplo,
la función de $3-$vectores definida por
$$f(x) =-2x_1+x_2-x_3+2.3$$
es afín, con $a = (-2,1,-1)$ y $b=2.3$.


Tanto las funciones afines como lineales son muy importantes en aplicaciones. Uno de sus principales usos es en la tarea de regresión de aprendizaje supervisado, una de las tareas mas utilizadas de inteligencia artificial.



*Representación matricial de funciones lineales*. Una función lineal $L : \mathbb{R}^n→\mathbb{R}^m$ se puede representar matricialmente como

$$L(x)=\begin{bmatrix}
a_{11} & a_{12} & \dots & a_{1n}\\
a_{21} & a_{22} & \dots & a_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
a_{m1} & a_{m2} & \dots & a_{mn}
\end{bmatrix}
\begin{bmatrix}x_1\\x_2\\\vdots\\x_n\end{bmatrix}=Ax=x_1a_1+x_2a_2+...+x_na_n.$$

donde $A$ es la matriz que representa a $L$ y $a_i$ es la $i-$esima columna de $A$. A las funciones $L$ se les suele llamar *transformaciones lineales*.

**Continuidad de funciones de varias variables**

En cálculo de una  variable motivamos la definición de continuidad de $f$ en $x$ preguntando
si el conocimiento aproximado de $x$ es suficiente para dar un conocimiento aproximado
de $f(x)$. Esta idea se mantiene en el caso de funciones de varias variables.

*Teorema*. Si $F : D ⊂ \mathbb{R}^n→\mathbb{R}^m$ es continua en $D$, entonces para toda sucesión
$X_1, X_2, . . . , X_k , . . .$
de puntos en $D$ que convergen a un punto $X$ en $D$, la sucesión
$F(X_1), F(X_2), . . . , F(X_k), . . .$
converge a $F(X)$.

*Teorema*. Una función $F : D ⊂ \mathbb{R}^n→\mathbb{R}^m$, denotada
$F(X) = (f_1(X), f_2(X), . . . , f_m(X))$,
es continua en $D$ si y solo si cada función componente $f_j : D ⊂ \mathbb{R}^n→\mathbb{R}^m$ es continua en $D$.

---

Por ejemplo la función $F(t)=(\cos t,\sin t, t)$ es continua en todo su dominio.

*Teorema*. Si $f : D ⊂ \mathbb{R}^n→\mathbb{R}$  y $g : D ⊂ \mathbb{R}^n→\mathbb{R}$  son continuas en D entonces

(a) $f +g $ es continua en $D$,

(b) $f g$ es continua en $D$,

(c) $\displaystyle \frac{f}{g}$ es continua en todo punto $X$ donde $g(X)\not=0$.

*Ejemplos*

Determine un conjunto $D$ en el dominio de la función dada donde esta sea continua:

1.   $\displaystyle f(x,y)=\frac{x^2+y^2}{x^2-y^2}$.
2.   $\displaystyle F(X)=-\frac{X}{||X||}$, $F:\mathbb{R}^n→\mathbb{R}^n$.
3. $\displaystyle f(x,y,z)=(\sin(x^2+y^2),e^{xz},\ln (yz))$
