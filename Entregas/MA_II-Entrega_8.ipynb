3. (24 puntos) **Mínimos cuadrados**. Considere el problema de mínimos cuadrados:
$$
\min_{x \in \mathbb{R}^n} \|Ax - b\|^2 \qquad (1)
$$
donde $A \in \mathbb{R}^{m \times n}$ tiene rango de columna completo (esto es equivalente a ser no singular en el caso $m=n$) y $b \in \mathbb{R}^m$. En particular, $m \geq n$.

3.1 Muestre que la función objetivo $$f(x)=\|Ax - b\|^2=(Ax-b)^\top(Ax-b)$$ puede expresarse como una función cuadrática:
$$
f(x) = \frac{1}{2}x^\top Px + q^\top x + r
$$
donde: $P = 2A^\top A$ es simétrica y definida positiva, $q = -2A^\top b$ y $r = b^\top b$.

La función $f$ es fuertemente convexa, esto garantiza la existencia de un único minimizador global.

3.2 **Condición de Optimalidad**.
Considere el caso particular en que $A=\begin{bmatrix}
a_{11} & a_{12} \\
a_{21} & a_{22}
\end{bmatrix}$ y $b=\begin{bmatrix}
b_1 \\
b_2
\end{bmatrix}$. Muestre que el gradiente de $f$ es:
$$
\nabla f(x) = Px + q = 2A^\top Ax - 2A^\top b,
$$
y entonces los puntos estacionarios satisfacen:
$$
\nabla f(x) = 0 \implies A^\top Ax = A^\top b
$$
Estas son las **ecuaciones normales** para el problema de mínimos cuadrados.
