2.2 *Ecualización de canales*. Supongamos que $u_1,..., u_m$ es una señal (serie de  tiempo) que se transmite
(por ejemplo por radio). Un receptor recibe la señal $y = c\ast u$, donde el $n-$vector $c$ se denomina respuesta de impulso del canal. (Consulte la sección 7.4 del texto de Boyd) En la mayoría de las aplicaciones, $n$ es pequeño, por ejemplo, menos de 10 y $m$ es mucho mayor. Un ecualizador es un $k-$vector $h$ que satisface $h\ast c \approx e_1$, el primer vector unitario de longitud $n + k - 1$. El receptor ecualiza la señal recibida $y$  convolucionándola con el ecualizador para obtener $z = h\ast y$.

  a. ¿Cómo están $z$ (la señal recibida ecualizada) y $u$ (la señal transmitida original) relacionados? Sugerencia. Recuerde que $h\ast(c\ast u) = (h\ast c)\ast u$.

  b. Ejemplo numérico. Genere una señal $u$ de longitud $m = 50$, con cada entrada definida de manera aleatoria con valores que  sean $-1$ o $+1$. Dibuje $u$ y $y = c\ast u$, con $c = (1, 0.7,0.3)$. También
dibuje la señal ecualizada $z =h \ast y$, con

$$h=(0.9,-0.5,0.5,-0.4,0.3,-0.3,0.2,-0.1).$$

```python
import numpy as np
import matplotlib.pyplot as plt

def convolution(a, b):
    n = len(a)
    m = len(b)
    result_size = n + m - 1
    result = np.zeros(result_size)

    for i in range(n):
        for j in range(m):
            result[i + j] += a[i] * b[j]

    return result

# Example signals
a = [1, 2, 3, 4, 3, 2, 1]  # Original signal
b = [-1, 0, 1]       # Filter/kernel

conv_result = convolution(a, b)

# Time axes
t_a = np.arange(len(a))
t_conv = np.arange(len(conv_result))

# Plotting
plt.figure(figsize=(10, 5))

plt.subplot(2, 1, 1)
plt.stem(t_a, a, basefmt=" ")
plt.title("Original Signal")
plt.xlabel("Sample")
plt.ylabel("Amplitude")

plt.subplot(2, 1, 2)
plt.stem(t_conv, conv_result, basefmt=" ")
plt.title("Convoluted Signal")
plt.xlabel("Sample")
plt.ylabel("Amplitude")

plt.tight_layout()
plt.show()
```
