1.3 *Jacobiano de una función, $F:\mathbb{R}^n\rightarrow\mathbb{R}^n$*. La matriz de derivadas de una función $F:\mathbb{R}^n\rightarrow\mathbb{R}^n$ dada por $F(x)=F(x_1,...,x_n)=(f_1(x),...,f_n(x))$ en un punto $P$ se puede escribir en forma matricial como:

$$DF(P)=\begin{bmatrix}
\nabla f_1(x)^T\\
\vdots\\
\nabla f_n(x)^T
\end{bmatrix}.$$

El determinante de $DF(P)$ se denomina *Jacobiano* de $F$ en $A$ y se denota

$$JF(P)=\det (DF(P)).$$

Considere la función $F(x,y)=(x+y^2,x^3+xy)$. Encuentre el Jacobiano de $F$ en el punto $P=(1,2)$.
